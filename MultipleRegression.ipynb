```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
```

Create Dataset

```python
data={"size":[700,900,1050,1050,1200,1300,1300,1500,1700,1700,2000,2200],
      "bhk":[1,1,1,2,2,2,3,3,3,4,4,4],
      "price":[1400000,1700000,1950000,2025000,2450000,2700000,2800000,3200000,3500000,3900000,4200000,4500000]}
```

```python
data
```

```python
print(type(data))
```

```python
df=pd.DataFrame(data)
```

```python
df
```

```python
plt.scatter(df["size"],df["bhk"])
```

```python
plt.scatter(df["size"],df["price"])
```

```python
plt.scatter(df["price"],df["bhk"],)
```

```python
df.corr()
```

Seaborn Visualization

```python
sns.pairplot(df)
```

Finalize Dependent and Independent Features

```python
X=df[["size","bhk"]]
```

```python
y=df["price"]
```

Split the Dataset For Training and Testing

```python
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.18,random_state=42)
```

Train the Model

```python
model=LinearRegression()
```

```python
model.fit(X_train,y_train)
```

View Model Parameters

```python
print("Intercept:",model.intercept_)
```

```python
print("Coefficient:",model.coef_)
```

Make The Predictions

```python
y_pred=model.predict(X_test)
```

```python
print("Predicted Value:",y_pred)
```

```python
print("Actual Value:",y_test.values)
```

Evaluat Model

```python
mse=mean_squared_error(y_test,y_pred)
```

```python
mse
```

```python
r2=r2_score(y_test,y_pred)
```

```python
r2
```

Predict The Price For New Flat

```python
a=[[1900,2]]
p=model.predict(a)
print("Predicted Price = ",p)
```
